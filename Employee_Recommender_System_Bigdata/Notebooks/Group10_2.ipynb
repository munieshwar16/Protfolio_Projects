```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, countDistinct, when, sum, datediff, hour, coalesce, min, max, avg ,lit,split
from pyspark.sql.window import Window
import pyspark.sql.functions as F

# Create a SparkSession
spark = SparkSession.builder.appName("LoginAnalysis").getOrCreate()

# Load the CSV data into a DataFrame
logon_df = spark.read.csv("/content/drive/MyDrive/DataSets/logon_info.csv", header=True, inferSchema=True)
logon_df= logon_df.withColumn("time", split(logon_df["date"], " ")[1])
logon_df = logon_df.withColumn("date", split(logon_df["date"], " ")[0])  # Extract date part
 # Extract time part

# Convert the date column to a datetime format
logon_df.printSchema()
logon_df.show()
log_df = logon_df.filter(col('activity') == 'Logon')

# Filter logoff activities
logof_df = logon_df.filter(col('activity') == 'Logoff')
# Show the first few rows of each DataFrame
print("Schema of logon DataFrame:")
log_df.printSchema()
print("Sample of logon data:")
log_df.show()

print("Schema of logoff DataFrame:")
logof_df.printSchema()

# Show a sample of data from logon and logoff DataFrames


print("Sample of logoff data:")
logof_df.show()
```

```python
# Function to get the shape of a DataFrame
def get_shape(df):
    num_rows = df.count()
    num_cols = len(df.columns)
    return num_rows, num_cols

# Getting the shape of each DataFrame
logon_shape = get_shape(logon_df)
log_df_shape = get_shape(log_df)
logof_df_shape = get_shape(logof_df)

# Printing the shapes
print("Shape of logon DataFrame: Rows = {}, Columns = {}".format(logon_shape[0], logon_shape[1]))
print("Shape of logon data DataFrame: Rows = {}, Columns = {}".format(log_df_shape[0], log_df_shape[1]))
print("Shape of logoff DataFrame: Rows = {}, Columns = {}".format(logof_df_shape[0], logof_df_shape[1]))
```

```python
lodf = log_df.drop(col('id'))
loof_df  = logof_df.drop(col('id'))
```

```python
lodf.show()
```

```python
# Rename the last but two columns
loof_df = loof_df.withColumnRenamed("date", "logoff_date") \
                                       .withColumnRenamed("activity", "logoff_activity") \
                                       .withColumnRenamed("date", "logoff_date") \
                                       .withColumnRenamed("time" ,"logoff_time")\
                                       .withColumnRenamed("user" ,'logoff_user')\
                                       .withColumnRenamed("pc" ,"loggoff_pc")


# Show the updated DataFrame



```

```python
left_inner_join_df = lodf.join(loof_df, on=[(lodf["date"] == loof_df["logoff_date"]) & (lodf["user"] == loof_df["logoff_user"]) & (lodf["pc"] == loof_df["loggoff_pc"])], how="inner")
```

```python
left_inner_join_df.show()

# Function to get the shape of a DataFrame
def get_shape(df):
    num_rows = df.count()
    num_cols = len(df.columns)
    return num_rows, num_cols

# Getting the shape of each DataFrame
logon_shape_final = get_shape(left_inner_join_df)



# Printing the shapes
print("Shape of logon DataFrame: Rows = {}, Columns = {}".format(logon_shape_final[0], logon_shape_final[1]))
```

```python
from pyspark.sql.functions import unix_timestamp, col, round

# Convert time strings to Unix timestamps
left_inner_join_df = left_inner_join_df.withColumn("login_timestamp", unix_timestamp("time", "HH:mm:ss"))
left_inner_join_df = left_inner_join_df.withColumn("logoff_timestamp", unix_timestamp("logoff_time", "HH:mm:ss"))

# Calculate duration in seconds
left_inner_join_df = left_inner_join_df.withColumn("duration_seconds", col("logoff_timestamp") - col("login_timestamp"))

# Calculate duration in hours and round to 2 decimal places
left_inner_join_df = left_inner_join_df.withColumn("duration_hours", round(col("duration_seconds") / 3600, 2))

# Show the resulting DataFrame
left_inner_join_df.show()
```

```python
import plotly.express as px
import pandas as pd

# Assuming df is your PySpark DataFrame


# Convert PySpark DataFrame to Pandas DataFrame
df_d = left_inner_join_df.toPandas()

# Aggregate data to count distinct PCs used by each user per day
pc_count_df = df_d.groupby(['date', 'user'])['pc'].nunique().reset_index()

# Bar plot
fig_bar = px.bar(pc_count_df, x='user', y='pc', color='user',
                 title='Number of PCs Used by User per Day',
                 labels={'user': 'User', 'pc': 'Number of PCs'},
                 hover_data=['date'])

# Show plot
fig_bar.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
from pyspark.sql import SparkSession

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("Login and Logoff Visualization") \
    .getOrCreate()

# Assuming df is your PySpark DataFrame
df_pd = left_inner_join_df.toPandas()

# Convert PySpark DataFrame to Pandas DataFrame



# Scatter plot
fig_scatter = px.scatter(df_pd, x='login_timestamp', y='logoff_timestamp', color='user',
                         title='Clustering on Users Login and Logoff Timings',
                         labels={'login_timestamp': 'Login Timestamp', 'logoff_timestamp': 'Logoff Timestamp'},
                         hover_data=['user'])

# Add diagonal line for reference
fig_scatter.add_trace(go.Scatter(x=[df_pd['login_timestamp'].min(), df_pd['logoff_timestamp'].max()],
                                 y=[df_pd['login_timestamp'].min(), df_pd['logoff_timestamp'].max()],
                                 mode='lines', name='Diagonal Line'))

# Show plot
fig_scatter.show()

```

```python
# Specify the path where you want to save the CSV file
output_path = "path_to_output_file.csv"

# Write the DataFrame to a CSV file
left_inner_join_df.write.csv(output_path, header=True)
```

```python
from pyspark.sql import functions as F

# Group by date and user, and calculate the sum of duration_hours
daily_duration_df = left_inner_join_df.groupBy("date", "user").agg(F.sum("duration_hours").alias("total_duration"))

# Group by user and calculate the average duration per day
average_duration_df = daily_duration_df.groupBy("user").agg(F.avg("total_duration").alias("average_duration_per_day"))

# Round the average duration to two decimal places
average_duration_df = average_duration_df.withColumn("average_duration_per_day", F.round("average_duration_per_day", 2))

# Show the average working duration of users per day
average_duration_df.show()
```

```python
# Specify the path where you want to save the CSV file
output_path = "avg.csv"

# Write the DataFrame to a CSV file
average_duration_df.write.csv(output_path, header=True)
```

```python
import pandas as pd
import plotly.graph_objs as go
from plotly.subplots import make_subplots

# Convert PySpark DataFrame to Pandas DataFrame
average_duration_pd = average_duration_df.toPandas()

# Sort DataFrame by average duration
average_duration_pd = average_duration_pd.sort_values(by='average_duration_per_day', ascending=False)

# Create a bar plot
fig = make_subplots(rows=1, cols=1)

# Add bar trace
fig.add_trace(go.Bar(x=average_duration_pd['user'], y=average_duration_pd['average_duration_per_day'], name='Average Duration'),
              row=1, col=1)

# Update layout for better visualization
fig.update_layout(title="Average Duration Per Day for Users",
                  xaxis_title="User",
                  yaxis_title="Average Duration",
                  xaxis=dict(title="User"),
                  yaxis=dict(title="Average Duration (hours)"))

# Show the plot
fig.show()
```
